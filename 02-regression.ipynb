```python

df = pd.read_csv("car_fuel_efficiency_2026.csv")
```

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Split dataset
n = len(df)
n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

np.random.seed(42)
idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]].copy()
df_val = df.iloc[idx[n_train:n_train + n_val]].copy()
df_test = df.iloc[idx[n_train + n_val:]].copy()

target = 'fuel_efficiency_mpg'

features = [
    'model_year',
    'engine_displacement',
    'horsepower',
    'vehicle_weight',
    'acceleration'
]

y_train = df_train[target].values
y_val = df_val[target].values

df_train_zero = df_train[features].fillna(0)
df_val_zero = df_val[features].fillna(0)

model_zero = LinearRegression()
model_zero.fit(df_train_zero, y_train)

y_pred_zero = model_zero.predict(df_val_zero)

rmse_zero = np.sqrt(mean_squared_error(y_val, y_pred_zero))

print("RMSE with zero:", round(rmse_zero, 3))

# Calculate mean using training data only
horsepower_mean = df_train['horsepower'].mean()

df_train_mean = df_train[features].fillna(horsepower_mean)
df_val_mean = df_val[features].fillna(horsepower_mean)

model_mean = LinearRegression()
model_mean.fit(df_train_mean, y_train)

y_pred_mean = model_mean.predict(df_val_mean)

rmse_mean = np.sqrt(mean_squared_error(y_val, y_pred_mean))

print("RMSE with mean:", round(rmse_mean, 3))
print("Zero imputation:", round(rmse_zero, 3))
print("Mean imputation:", round(rmse_mean, 3))
```

```python
from sklearn.linear_model import Ridge

# Features and target
features = [
    'model_year',
    'engine_displacement',
    'horsepower',
    'vehicle_weight',
    'acceleration'
]

target = 'fuel_efficiency_mpg'

# Fill missing values with 0
X_train = df_train[features].fillna(0)
X_val = df_val[features].fillna(0)

y_train = df_train[target].values
y_val = df_val[target].values

# Try different regularization strengths
r_values = [0, 0.01, 0.1, 1, 5, 10, 100]

scores = []

for r in r_values:
    model = Ridge(alpha=r)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_val)
    rmse = np.sqrt(mean_squared_error(y_val, y_pred))

    scores.append((r, round(rmse, 4)))
    print(f"r = {r:<6} RMSE = {rmse:.4f}")

# Select the smallest r if RMSE scores tie
best_r, best_rmse = min(scores, key=lambda x: (x[1], x[0]))

print(f"\nBest r: {best_r}")
print(f"Best RMSE: {best_rmse:.4f}")
```

```python
features = [
    'model_year',
    'engine_displacement',
    'horsepower',
    'vehicle_weight',
    'acceleration'
]

target = 'fuel_efficiency_mpg'

n = len(df)
n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

scores = []

for seed in range(10):
    np.random.seed(seed)
    idx = np.arange(n)
    np.random.shuffle(idx)

    df_train = df.iloc[idx[:n_train]].copy()
    df_val = df.iloc[idx[n_train:n_train + n_val]].copy()
    df_test = df.iloc[idx[n_train + n_val:]].copy()

    # Fill missing values with 0
    X_train = df_train[features].fillna(0)
    X_val = df_val[features].fillna(0)

    y_train = df_train[target].values
    y_val = df_val[target].values

    # Train linear regression without regularization
    model = LinearRegression()
    model.fit(X_train, y_train)

    # Calculate validation RMSE
    y_pred = model.predict(X_val)
    rmse = np.sqrt(mean_squared_error(y_val, y_pred))

    scores.append(rmse)
    print(f"Seed {seed}: RMSE = {rmse:.4f}")

# Calculate standard deviation
std = np.std(scores)
print("Standard deviation:", round(std, 3))
```

```python
# Split dataset using seed 9
n = len(df)
n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

np.random.seed(9)
idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]].copy()
df_val = df.iloc[idx[n_train:n_train + n_val]].copy()
df_test = df.iloc[idx[n_train + n_val:]].copy()

# Combine training and validation datasets
df_train_full = __import__('pandas').concat(
    [df_train, df_val]
)

# Select features and target
features = [
    'model_year',
    'engine_displacement',
    'horsepower',
    'vehicle_weight',
    'acceleration'
]

target = 'fuel_efficiency_mpg'

# Fill missing values with 0
X_train = df_train_full[features].fillna(0)
y_train = df_train_full[target]

X_test = df_test[features].fillna(0)
y_test = df_test[target]

# Train Ridge Regression with r = 0.001
model = Ridge(alpha=0.001)
model.fit(X_train, y_train)

# Evaluate on test dataset
y_pred = model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("Test RMSE:", round(rmse, 3))
```
